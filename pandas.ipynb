```python
data_with_missing = {
    "cars": ["BMW M3 F80 CS", "Volvo", None, "Tesla", "Ferrari", None, "Ford"],
    "counts": [3, 7, None, 5, 6, None, 10],
}

df_missing = pd.DataFrame(data_with_missing)
print(df_missing)
```

```python
df_missing.isnull()
```

```python
df_missing.isnull().sum()
```

```python
df_missing.dropna() # unhealthy bo'lsa drop qilarkan
```

```python
df_missing.dropna(how="all")
```

```python
df_missing.dropna(axis=1)
```

```python
df_missing.dropna(subset=['counts'])
```

```python
df_missing.fillna(0)
```

```python
df_missing.fillna("unknown")
```

```python
df_missing.fillna({"cars": "unknown", "counts": 0})
```
